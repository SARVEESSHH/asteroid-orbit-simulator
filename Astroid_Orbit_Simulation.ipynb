```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 1000)

# Earth
earth_r = 1
earth_x = earth_r * np.cos(theta)
earth_y = earth_r * np.sin(theta)

# Ceres
a_ceres = 2.77
e_ceres = 0.08

r_ceres = a_ceres * (1 - e_ceres**2) / (1 + e_ceres * np.cos(theta))
x_ceres = r_ceres * np.cos(theta)
y_ceres = r_ceres * np.sin(theta)

# Vesta
a_vesta = 2.36
e_vesta = 0.09

r_vesta = a_vesta * (1 - e_vesta**2) / (1 + e_vesta * np.cos(theta))
x_vesta = r_vesta * np.cos(theta)
y_vesta = r_vesta * np.sin(theta)

plt.figure(figsize=(8,8))

plt.plot(earth_x, earth_y, label="Earth")
plt.plot(x_ceres, y_ceres, label="Ceres")
plt.plot(x_vesta, y_vesta, label="Vesta")

plt.scatter(0, 0, s=300, label="Sun")

plt.axis("equal")
plt.title("Earth, Ceres, and Vesta Orbits")
plt.xlabel("Distance (AU)")
plt.ylabel("Distance (AU)")
plt.legend()

plt.savefig("solar_system.png", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Ceres orbital parameters
a = 2.77  # Semi-major axis (AU)
e = 0.08  # Eccentricity

theta = np.linspace(0, 2*np.pi, 2000)

# Orbit equation
r = a * (1 - e**2) / (1 + e * np.cos(theta))

x = r * np.cos(theta)
y = r * np.sin(theta)

plt.figure(figsize=(8,8))

# Orbit
plt.plot(x, y, linewidth=2, label="Ceres Orbit")

# Sun at focus
plt.scatter(0, 0, s=300, marker="o", label="Sun")

# Current asteroid position
idx = 400
plt.scatter(x[idx], y[idx], s=100, label="Ceres")

plt.title("Orbital Simulation of Asteroid 1 Ceres")
plt.xlabel("Distance (AU)")
plt.ylabel("Distance (AU)")
plt.axis("equal")
plt.grid(True)
plt.legend()

plt.savefig("orbit_plot.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# ------------------------
# Orbital Parameters
# ------------------------

theta = np.linspace(0, 2*np.pi, 1000)

# Earth
a_earth = 1.00
e_earth = 0.0167
T_earth = 1.00

# Vesta
a_vesta = 2.36
e_vesta = 0.09
T_vesta = 3.63

# Ceres
a_ceres = 2.77
e_ceres = 0.08
T_ceres = 4.61

# ------------------------
# Orbit Coordinates
# ------------------------

r_earth = a_earth*(1-e_earth**2)/(1+e_earth*np.cos(theta))
x_earth = r_earth*np.cos(theta)
y_earth = r_earth*np.sin(theta)

r_vesta = a_vesta*(1-e_vesta**2)/(1+e_vesta*np.cos(theta))
x_vesta = r_vesta*np.cos(theta)
y_vesta = r_vesta*np.sin(theta)

r_ceres = a_ceres*(1-e_ceres**2)/(1+e_ceres*np.cos(theta))
x_ceres = r_ceres*np.cos(theta)
y_ceres = r_ceres*np.sin(theta)

# ------------------------
# Figure
# ------------------------

fig, ax = plt.subplots(figsize=(8,8))

ax.plot(x_earth, y_earth, label="Earth")
ax.plot(x_vesta, y_vesta, label="Vesta")
ax.plot(x_ceres, y_ceres, label="Ceres")

ax.scatter(0, 0, s=300, label="Sun")

earth_dot, = ax.plot([], [], marker='o', markersize=8, label="Earth Position")
vesta_dot, = ax.plot([], [], marker='o', markersize=8, label="Vesta Position")
ceres_dot, = ax.plot([], [], marker='o', markersize=8, label="Ceres Position")

ax.set_aspect("equal")
ax.grid(True)
ax.legend()
ax.set_title("Earth, Vesta and Ceres Orbit Simulation")

# ------------------------
# Animation
# ------------------------

frames = 1000

def update(frame):

    earth_index = int(frame * (4.61 / T_earth)) % len(theta)
    vesta_index = int(frame * (4.61 / T_vesta)) % len(theta)
    ceres_index = int(frame * (4.61 / T_ceres)) % len(theta)

    earth_dot.set_data(
        [x_earth[earth_index]],
        [y_earth[earth_index]]
    )

    vesta_dot.set_data(
        [x_vesta[vesta_index]],
        [y_vesta[vesta_index]]
    )

    ceres_dot.set_data(
        [x_ceres[ceres_index]],
        [y_ceres[ceres_index]]
    )

    return earth_dot, vesta_dot, ceres_dot

ani = FuncAnimation(
    fig,
    update,
    frames=frames,
    interval=20,
    blit=True
)

ani.save(
    "orbit_animation.gif",
    writer="pillow",
    fps=30
)

plt.close()

print("Saved: orbit_animation.gif")
```
